```python
import matplotlib.pyplot as plt
import numpy as np
```

Plotting Accuracy

```python
def plot_accuracy(models, accuracy, dataset):
    xpoints = np.array(models)
    ypoints = np.array(accuracy)

    plt.ylim(0, 100)
    # show the values on the plot
    for x, y in zip(xpoints, ypoints):
        label = "{:.2f}%".format(y)
        plt.annotate(label, (x, y), textcoords="offset points", xytext=(0, 10), ha='center')
        
    plt.bar(xpoints, ypoints, color=['red', 'green', 'blue'], width=0.5)
    plt.plot(xpoints, ypoints, linestyle='dashed', marker='*', color='black')
    plt.title(f"Accuracy of different models on {dataset}")
    plt.xlabel("Model")
    plt.ylabel("Accuracy (%)")

    # save file as eps for latex
    plt.savefig(f"accuracy-{dataset}.eps", format="eps", dpi=1200)
```

```python
def plot_time(models, times, dataset, limit = 60):
    xpoints = np.array(models)
    ypoints = np.array(times)

    for x, y in zip(xpoints, ypoints):
        label = f"{y} mins" if y < 60 else f"{y // 60} hrs {y % 60} mins"
        plt.annotate(label, (x, y), textcoords="offset points", xytext=(0, 10), ha='center')
    
    plt.ylim(0, limit)
    plt.bar(xpoints, ypoints, color=['red', 'green', 'blue'], width=0.5)
    plt.plot(xpoints, ypoints, linestyle='dashed', marker='*', color='black')
    plt.title(f"Training Time of models on {dataset}")
    plt.xlabel("Model")
    plt.ylabel("Training Time (in minutes)")

    # save file as eps
    plt.savefig(f'training-time-{dataset}.eps', format="eps", dpi=1200)
```

```python
models = ["CNN", "GPT-Neo (125M)", "OPT (125M)"]
```

```python
plot_accuracy(models, [59.06, 83.24, 85.29], "dataset-1")
```

```python
plot_time(models, [6, 37, 34], "dataset-1")
```

```python
plot_accuracy(models, [68.3, 90.61, 96.72], "dataset-2")
```

```python
plot_time(models, [276, 159, 151], "dataset-2", limit=300)
```
